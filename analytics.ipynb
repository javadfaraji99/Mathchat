```python
def analyze_json_keys(base_folder, key_list):
    all_values = {key: [] for key in key_list}
    
    for category, category_folder in math_type_mapping.items():
        folder_path = os.path.join(base_folder, category_folder)
        if not os.path.exists(folder_path):
            continue
        
        values = {key: [] for key in key_list}
        
        for file_name in os.listdir(folder_path):
            if file_name.startswith('question_') and file_name.endswith('.json'):
                file_path = os.path.join(folder_path, file_name)
                with open(file_path, 'r') as f:
                    data = json.load(f)
                
                for key in key_list:
                    if key in data:
                        values[key].append(data[key])
                        all_values[key].append(data[key])
        
        print(f"\n{category}:")
        for key in key_list:
            if values[key]:
                print(f"  {key}: min={np.min(values[key])}, max={np.max(values[key])}, mean={np.mean(values[key]):.2f}")
            else:
                print(f"  {key}: No data")
    
    print(f"\nALL CATEGORIES COMBINED:")
    for key in key_list:
        if all_values[key]:
            print(f"  {key}: min={np.min(all_values[key])}, max={np.max(all_values[key])}, mean={np.mean(all_values[key]):.2f}")
        else:
            print(f"  {key}: No data")
```

```python

def analyze_discrete_values(base_folder, key, discrete=1, ranges=None):
    all_values = []
    
    for category, category_folder in math_type_mapping.items():
        folder_path = os.path.join(base_folder, category_folder)
        if not os.path.exists(folder_path):
            continue
        
        values = []
        
        for file_name in os.listdir(folder_path):
            if file_name.startswith('question_') and file_name.endswith('.json'):
                file_path = os.path.join(folder_path, file_name)
                with open(file_path, 'r') as f:
                    data = json.load(f)
                
                if key in data:
                    values.append(data[key])
                    all_values.append(data[key])
        
        print(f"\n{category}:")
        
        if not values:
            print("  No data")
            continue
        
        if discrete == 1:
            counter = Counter(values)
            print(f"  {key} value counts:")
            for val, count in sorted(counter.items()):
                print(f"    {val}: {count}")
        else:
            if not ranges:
                print("  No ranges provided")
                continue
            print(f"  {key} range counts:")
            for start, end in ranges:
                count = sum(1 for v in values if start <= v <= end)
                print(f"    [{start}, {end}]: {count}")
    
    print(f"\nALL CATEGORIES COMBINED:")
    
    if not all_values:
        print("  No data")
        return
    
    if discrete == 1:
        counter = Counter(all_values)
        print(f"  {key} value counts:")
        for val, count in sorted(counter.items()):
            print(f"    {val}: {count}")
    else:
        if not ranges:
            print("  No ranges provided")
            return
        print(f"  {key} range counts:")
        for start, end in ranges:
            count = sum(1 for v in all_values if start <= v <= end)
            print(f"    [{start}, {end}]: {count}")
```

```python
def analyze_python_usage(base_folder):
    results = {}
    
    for category, category_folder in math_type_mapping.items():
        folder_path = os.path.join(base_folder, category_folder)
        if not os.path.exists(folder_path):
            continue
        
        python_used = []
        no_python = []
        invalid_queries = []
        
        for file_name in os.listdir(folder_path):
            if file_name.startswith('question_') and file_name.endswith('.json'):
                file_path = os.path.join(folder_path, file_name)
                with open(file_path, 'r') as f:
                    data = json.load(f)
                
                q_id = data.get('problem_id', file_name.replace('.json', ''))
                
                has_python = False
                if 'messages' in data:
                    for msg in data['messages']:
                        if msg['role'] == 'assistant' and 'content' in msg:
                            content = msg['content']
                            if 'Python' in content or 'python' in content or 'print(' in content:
                                has_python = True
                                break
                
                valid_q_count = data.get('valid_q_count', 0)
                total_q_count = data.get('total_q_count', 0)
                has_invalid = valid_q_count < total_q_count and total_q_count > 0
                
                if has_python:
                    python_used.append(q_id)
                else:
                    no_python.append(q_id)
                
                if has_invalid:
                    invalid_queries.append(q_id)
        
        results[category] = {
            'python_used': python_used,
            'no_python': no_python,
            'invalid_queries': invalid_queries,
            'total': len(python_used) + len(no_python)
        }
        
        print(f"\n{category}:")
        print(f"  Total questions: {len(python_used) + len(no_python)}")
        print(f"  Python used: {len(python_used)}")
        print(f"  No Python: {len(no_python)}")
        print(f"  Invalid queries: {len(invalid_queries)}")
    
    total_python = sum(len(r['python_used']) for r in results.values())
    total_no_python = sum(len(r['no_python']) for r in results.values())
    total_invalid = sum(len(r['invalid_queries']) for r in results.values())
    total_all = sum(r['total'] for r in results.values())
    
    print(f"\nALL CATEGORIES COMBINED:")
    print(f"  Total questions: {total_all}")
    print(f"  Python used: {total_python}")
    print(f"  No Python: {total_no_python}")
    print(f"  Invalid queries: {total_invalid}")
    
    output_file = "python_usage_analysis.json"
    with open(output_file, 'w') as f:
        json.dump({
            'categories': results,
            'total': {
                'total_questions': total_all,
                'python_used': total_python,
                'no_python': total_no_python,
                'invalid_queries': total_invalid
            }
        }, f, indent=2)
    
    print(f"\nResults saved to: {output_file}")
```

```python
analyze_python_usage(base_folder=base_folder)
```

```python
def find_errors(root_folder, output_file="errors.json"):
    errors_dict = {}
    pattern = "Error:"
    
    for dirpath, dirnames, filenames in os.walk(root_folder):
        for filename in filenames:
            if filename.endswith(".json"):
                filepath = os.path.join(dirpath, filename)
                try:
                    with open(filepath, "r", encoding="utf-8") as f:
                        data = json.load(f)
                    
                    messages = data.get("messages", [])
                    file_errors = []
                    for msg in messages:
                        if msg.get("role") == "user" and pattern in msg.get("content", ""):
                            file_errors.append(msg["content"])
                    
                    if file_errors:
                        errors_dict[filename[:-5]] = file_errors
                except (json.JSONDecodeError, KeyError, IOError):
                    continue
    
    with open(output_file, "w", encoding="utf-8") as f:
        json.dump(errors_dict, f, indent=2)
    
    return errors_dict
```

```python
def compare_invalid_queries(categories_file, errors_file):
    with open(categories_file, "r", encoding="utf-8") as f:
        categories = json.load(f)
    with open(errors_file, "r", encoding="utf-8") as f:
        errors = json.load(f)
    
    error_keys = set(errors.keys())
    category_keys = set()
    
    for key, value in categories.get("categories", {}).items():
        if isinstance(value, dict):
            invalid = value.get("invalid_queries", [])
            if isinstance(invalid, list):
                for item in invalid:
                    if isinstance(item, str):
                        category_keys.add(item)
                    elif isinstance(item, dict):
                        for k in item.keys():
                            category_keys.add(k)
            elif isinstance(invalid, dict):
                for k in invalid.keys():
                    category_keys.add(k)

    not_in_both = error_keys.symmetric_difference(category_keys)
    
    for name in sorted(not_in_both):
        print(name)
```
